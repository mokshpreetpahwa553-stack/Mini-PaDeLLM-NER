```python
def count_mentions_by_label(examples):
    counts = Counter()
    for example in examples:
        for label, mention_text in example["mentions"]:
            counts[label] += 1
    return counts
```

```python
print(count_mentions_by_label(train))
print(count_mentions_by_label(validation))
print(count_mentions_by_label(test))
```
